# Ejemplo de uso de un LLM

Este es un ejemplo de cómo usar un modelo de lenguaje grande (LLM) para generar texto a partir de un prompt dado.
En este caso, utilizaremos el modelo [distilgpt2](https://huggingface.co/distilbert/distilgpt2) de Hugging Face.

# Instalación de dependencias

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

# Detección de dispositivo
*¿El entorno de ejecución tiene una GPU disponible?*
Si es así, usaremos la GPU para acelerar el proceso de generación de texto.
De lo contrario, usaremos la CPU.


```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)
```

# Configuración del modelo y el tokenizador
En esta sección, cargaremos el modelo y el tokenizador de Hugging Face.
Asegúrate de que el modelo que elijas sea compatible con tu entorno de ejecución (GPU o CPU).

```python
MODEL_NAME = "distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME).to(device)
```

# Usa el modelo para generar texto
Ahora que tenemos el modelo y el tokenizador configurados, podemos usar el modelo para generar texto a partir de un prompt dado.

```python
PROMPT = "Hello, I'm a language model"

inputs = tokenizer(PROMPT, return_tensors="pt").to(device)
output = model.generate(
    **inputs,
    max_length=50,
    do_sample=True,
    temperature=0.8
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```
