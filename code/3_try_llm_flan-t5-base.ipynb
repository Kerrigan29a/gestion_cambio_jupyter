# Ejemplo de uso de un LLM

Este es un ejemplo de cómo usar un modelo de lenguaje grande (LLM) para generar texto a partir de un prompt dado.
En este caso, utilizaremos el modelo [flan-t5-base](https://huggingface.co/google/flan-t5-base) de Hugging Face.

# Instalación de dependencias

```python
import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
```

# Detección de dispositivo
*¿El entorno de ejecución tiene una GPU disponible?*
Si es así, usaremos la GPU para acelerar el proceso de generación de texto.
De lo contrario, usaremos la CPU.


```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)
```

# Configuración del modelo
En esta sección, cargaremos el modelo y el tokenizador de Hugging Face.
Asegúrate de que el modelo que elijas sea compatible con tu entorno de ejecución (GPU o CPU).

```python
MODEL_NAME = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

if device == "cuda":
    model = model.to(device)

print(f"Modelo cargado: {MODEL_NAME}")
```

# Usa el modelo para una instrucción simple
En este ejemplo usaremos una instrucción breve y determinista para validar que el modelo quedó cargado y puede generar una respuesta correctamente.

```python
PROMPT = "Translate from Spanish to English: En un lugar de La Mancha, de cuyo nombre no quiero acordarme, ..."

inputs = tokenizer(PROMPT, return_tensors="pt")
if device == "cuda":
    inputs = {name: tensor.to(device) for name, tensor in inputs.items()}

with torch.no_grad():
    output_tokens = model.generate(
        **inputs,
        max_new_tokens=40,
        do_sample=False,
    )

response = tokenizer.decode(output_tokens[0], skip_special_tokens=True)
print(response)
```
